```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

### Example

```python
d = stats.norm.rvs(loc = 5.0, scale = .01, size = 1000000)
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
x = np.linspace(4.95,5.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()
```

```python
my_d = stats.norm.rvs(loc = -10.0, scale = .05, size = 1000000)
```

* my_d = stats.norm.rvs(loc = -10.0, scale = .5, size = 1000000)
    * Defines my_d with a normal/gaussian sample of 1000000 numbers with a central value of -10.0 and standard deviation of 0.05
    * stats.norm.rsv(loc = central value or mean, scale = standard deviation, size = number of random values drawn from a gaussian distribution)
        * norm draws from the equation $ f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^\frac{-(x-\mu)^2}{2\sigma^2} $

```python
fig, ax = plt.subplots(1, 1)
ax.hist(my_d, 50, density = True)
plt.tick_params(labelsize = 24)
plt.xlim([-10.25, -9.75])
x = np.linspace(-10.25, -9.75, 1000)
ax.plot(x, stats.norm.pdf(x, loc = -10.0, scale = 0.05), linewidth = 8, alpha = 0.7)
plt.show()
```

* fig, ax = plt.subplots(1, 1)
    * Creates a figure with one subplot on it (1 row, 1 column)
    * fig, ax = plt.subplots(# of subplots in row, # of subplots in column) 
* ax.hist(my_d, 50, density=True)
    * Creates a probability density histogram plotting my_d (my gaussian distribution) into 50 bins
    * ax.hist(x = my data, bins = # of bins, density=True)
        * Density = True returns probability density where density = (# of bins) / ((# of total counts) * (bin width)) so that the area under the histogram integrates to 1   
* plt.tick_params(labelsize = 24)
    * Changes the fontsize of the labels for the tick marks to size 24
    * plt.tick_params(labelsize = size)
* plt.xlim([-10.25,-9.75])
    * Changes the lower limit of x-axis to -10.25 and the upper limit of x-axis to -9.75
    * xlim([left = lower limit to x-axis, right = upper limit to x-axis])
* x = np.linspace(-10.25 , -9.75, 1000)
    * Defines x as an array of 1000 numbers evenly spaced between -10.25 and -9.75
    * numpy.linspace(start = start value, stop = stop value, num = number of evenly spaced samples between [start, stop])
* ax.plot(x, stats.norm.pdf(x, loc = -10.0, scale = 0.05), linewidth = 8, alpha = 0.7)
    * Plots curve from defined x and y from stats.norm.pdf with a linewidth of 8 and an opacity of 0.7
    * ax.plot(x, y, linewidth = width of line plotted, alpha = opacity of line plotted)
    * stats.norm.pdf(x, loc = -10.0, scale = 0.05)
        * Plots a probability density function from the defined x with a central value of -10 and standard deviation of 0.05 (same as my_d)
        * stats.norm.pdf(x, loc = central value or mean, scale = standard deviation)
* plt.show()
    * Displays figure

* plt.yscale('log')
    * Sets y-axis's scale to a log scale 
    * plt.yscale(value)

Plots the data on log scale for the y-axis. Taking the log is good for telling if the data follows a Gaussian distribution as, if it does, it will have a parabolic shape. That is the case here.

```python
fig, ax = plt.subplots(1, 1)
ax.hist(my_d, 50, density = True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([-10.25, -9.75])
x = np.linspace(-10.25, -9.75, 1000)
ax.plot(x, stats.norm.pdf(x, loc = -10.0, scale = 0.05), linewidth = 8, alpha = 0.7)
plt.show()
```
