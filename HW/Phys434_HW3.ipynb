### Question 1
We're going to briefly look at truth tables. As an example let's assume we have a 100 photos, some of birds and some of people. In a truth table you assume you can categorize the true answer—let's say you pay an undergraduate to sort the photos into birds and people. Then you want to have an automated sorting algorithm. However that works, whether it is by looking a the color at the center of photo or using a neural network (NN) or an AI agent, it will also sort the photos but imperfectly. Given two sorts you can then arrange a truth table:

```python
data_NN = {
    " ": ["NN Bird", "NN Person"],
    "True Bird": [45, 3],
    "True Person": [5, 47]
}

df_NN = pd.DataFrame(data_NN)
df_NN
```

There are 48 birds and 52 People (columns), and there are 8 miscategorizations. The AI called three of the birds people, and 5 of the people birds.

This is useful because it not only shows the number of errors, but the type, and not all errors are created equal. Let's say our identification system is being used to keep birds from escaping the aviary at the zoo by locking the aviary door when a bird is trying to escape. We really don't want to let birds escape, but locking a person in for 30 extra seconds is not a big deal. In this case false bird identification is not so bad, but false person identification lets a bird escape.

Let's say the above truth table is the current system, and you've developed a new algorithm (NA) with the following truth table:

```python
data_NA = {
    " ": ["N Bird", "NA Person"],
    "True Bird": [47, 1],
    "True Person": [11, 42]
}

df_NA = pd.DataFrame(data_NA)
df_NA
```

#### Part (a)
What is the pdf of the sum of two identical exponential distributions?

$ \Large f_Z(z)$ 

$ \Large = \int_{-\infty}^{\infty} f_{X_1}(x_1)f_{X_2}(z-x_1)dx_1$

$ \Large = \int_0^z \lambda_1 e^{-\lambda_1 x_1} \lambda_2 e^{-\lambda_2(z-x_1)}dx_1$

$ \Large = \lambda_1 \lambda_2 e^{-\lambda_2z}\int_0^z e^{(\lambda_2 - \lambda_1)x_1} dx_1 $

$ \Large = \lambda^2 z e^{-\lambda z} $ where $\lambda = \lambda_1 = \lambda_2 $

(From Wikipedia: Exponential Distribution - Sum of two independent exponential random variables)

#### Part (b)
What is the pdf of the ratio of two zero-mean unity variance normal distributions $X_1/X_2$

If $X_1$ and $X_2$ are two zero-mean unity variance normal distributions ($X \sim N(0,1)$), then their ratio, $X_1 / X_2$, follows a standard Cauchy distribution ($X_1 / X_2 \sim Cauchy(0, 1)$)
